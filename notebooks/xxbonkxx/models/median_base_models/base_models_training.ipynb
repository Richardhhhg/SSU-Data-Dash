```python
import pandas as pd
import numpy as np
import sys
from pathlib import Path
import torch # Torch needs to be installed manually if you want cuda. See installation command on pytorch website

DIR_ROOT = Path.cwd().parent.parent.parent
sys.path.insert(0, str(DIR_ROOT))

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

%load_ext autoreload
%autoreload 2
```

```python
SEED = 987
VAL_START = pd.Timestamp("2025-12-01")    # sales on or after this date are validation
```

```python
import data

listings = data.load("listings")

train = listings[listings["date_sold"] < VAL_START]
val = listings[listings["date_sold"] >= VAL_START]

print(f"train: {len(train):,} sales up to {VAL_START.date()}")
print(f"val:   {len(val):,} sales from {VAL_START.date()}")
```

```python
from base_models import BaseModelConfig, run_base_model

# name, column in listings to scale by, columns to drop from the features
CONFIGS = [
    BaseModelConfig("style", "house_style", ["house_style"], n_trials=10, debug=True),
    BaseModelConfig("type", "house_type_name", ["house_type"], n_trials=10, debug=True),
    # BaseModelConfig("community", "community", ["community"], n_trials=50, debug=True),
    BaseModelConfig("municipality", "municipality", ["municipality"], n_trials=10, debug=True),
]
```

```python
results = {cfg.name: run_base_model(cfg, train, val, device=DEVICE, seed=SEED) for cfg in CONFIGS}
```

```python
# Inputs for the meta model: one column of log price predictions per base model.
# Rows in the first fold block have no OOF prediction, so drop them before training on oof_train.
oof_train = pd.concat([r["oof_train"] for r in results.values()], axis=1).dropna()
pred_val = pd.concat([r["pred_val"] for r in results.values()], axis=1)

for name, r in results.items():
    print(name, r["best_params"])
oof_train.head()
```
