## What Do the Features and the Data Mean:
### Different Datasets:
- listings: this includes all the available features from `off market` and `list aware` + some extra.
- Off Market Price: a dataset to predict sale price of houses when not given the list price
- List Aware Price: given the list price, predict the sale price of the house

For both the `off market` dataset and the `list aware` dataset, you are predicting the sales price as the target.

### Features
Features in All Datasets: ml_num, remarks, bedrooms, date_listed, washrooms, parking_total, lot_front, house_style, lot_depth, longitude, latitude, community, garage, house_type_name, house_area, municipality
Features that Don't Appear in Test Splits: external_avm_estimate, days_on_market, external_avm_estimate_date, date_sold

Meaning of Features:
- ml_num: the unique identifier for the observation
- remarks: notes about the house in the listing (written by an AI...)
- bedrooms: number of bedrooms (0-6)
- date_listed: Just the date listed...
- washrooms: Number of washrooms (1 - 6)
- parking_total: number of parking spots (0-6)
- lot_front: width of the property (ft)
- house_style: what style the house is (16 unique)
- lot_depth: how deep the property is from front to back (ft)
- longitude: east west coordinates
- latitude: north south coordinates
- community: what community the house is in (145 unique)
- garage: number of garages (0-6)
- house_type_name: what kind of house it is (6 unique)
- house_area: are of the house (sq ft, sometimes a range)
- municipality: what municipality house is in (5 unique)

Meaning of Listings Only Features:
- external_avm_estimate: the price estimated by a 3rd party
- days_on_market: how long the house has been on the market
- external_avm_estimate_date: the date of the 3rd party estimate, only informative if it is before the selling date
- date_sold: when the house was sold

#### Additional Considerations for Features
- for the categorical features (house_type_name, municipality, house_style, community) consider whether or not these features may be ordinal, ie. is one house style objectively better than another in some way.
